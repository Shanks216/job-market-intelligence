```python
import pandas as pd
import numpy as np

file_path = "../data/indian-job-market-dataset-2025.xlsx"

df = pd.read_excel(file_path)

print("Raw shape:", df.shape)
```

```python
df = df.rename(columns={
    "jobId": "job_id",
    "companyName": "company_name",
    "jobUploaded": "job_uploaded",
    "tagsAndSkills": "tags_and_skills",
    "companyId": "company_id",
    "ReviewsCount": "reviews_count",
    "AggregateRating": "aggregate_rating",
    "jobDescription": "job_description",
    "minimumSalary": "minimum_salary",
    "maximumSalary": "maximum_salary",
    "minimumExperience": "minimum_experience",
    "maximumExperience": "maximum_experience"
})

df.columns.tolist()
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate job IDs:", df["job_id"].duplicated().sum())
```

```python
df = df.drop_duplicates(
    subset="job_id",
    keep="first"
).copy()

print("After deduplication:", df.shape)
```

```python
text_columns = [
    "title",
    "company_name",
    "currency",
    "location",
    "tags_and_skills",
    "experience",
    "salary",
    "job_description"
]

for col in text_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
    )
```

```python
for col in text_columns:
    df[col] = df[col].replace(
        ["", "nan", "None", "NA", "N/A"],
        pd.NA
    )
```

```python
import re

def normalize_title(title):

    if pd.isna(title):
        return pd.NA

    title = str(title).lower().strip()

    # Replace punctuation with spaces
    title = re.sub(r"[^a-z0-9+#./& -]", " ", title)

    # Common abbreviations
    replacements = {
        r"\bsr\b": "senior",
        r"\bsr\.\b": "senior",
        r"\bjr\b": "junior",
        r"\bjr\.\b": "junior",
        r"\bdev\b": "developer",
        r"\beng\b": "engineer",
        r"\bmgmt\b": "management",
        r"\bdept\b": "department",
        r"\bassoc\b": "associate",
        r"\bvp\b": "vice president",
        r"\bdir\b": "director"
    }

    for pattern, replacement in replacements.items():
        title = re.sub(
            pattern,
            replacement,
            title
        )

    # Normalize whitespace
    title = re.sub(r"\s+", " ", title).strip()

    return title.title()
```

```python
df["normalized_title"] = df["title"].apply(
    normalize_title
)
```

```python
df[
    ["title", "normalized_title"]
].head(30)
```

```python
salary_columns = [
    "minimum_salary",
    "maximum_salary"
]

for col in salary_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

    df.loc[
        df[col] <= 0,
        col
    ] = np.nan
```

```python
df[
    ["salary", "minimum_salary", "maximum_salary"]
].head(20)
```

```python
df["salary_midpoint"] = (
    df["minimum_salary"] +
    df["maximum_salary"]
) / 2
```

```python
df["salary_midpoint"].describe()
```

```python
df["currency"].value_counts(dropna=False)
```

```python
experience_columns = [
    "minimum_experience",
    "maximum_experience"
]

for col in experience_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )
```

```python
df["experience_midpoint"] = (
    df["minimum_experience"] +
    df["maximum_experience"]
) / 2
```

```python
df["job_uploaded"] = pd.to_datetime(
    df["job_uploaded"],
    errors="coerce"
)
```

```python
print(df["job_uploaded"].dtype)

print(
    df["job_uploaded"].min(),
    "→",
    df["job_uploaded"].max()
)
```

```python
df["posting_year"] = (
    df["job_uploaded"].dt.year
)

df["posting_month"] = (
    df["job_uploaded"].dt.to_period("M")
    .astype(str)
)
```

```python
df["location"] = (
    df["location"]
    .astype("string")
    .str.strip()
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)
```
