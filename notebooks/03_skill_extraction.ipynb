```python
import spacy
from spacy.matcher import PhraseMatcher

nlp = spacy.blank("en")

matcher = PhraseMatcher(
    nlp.vocab,
    attr="LOWER"
)

patterns = [
    nlp.make_doc(skill)
    for skill in skills
]

matcher.add(
    "SKILLS",
    patterns
)

print("PhraseMatcher ready")
```

```python
from bs4 import BeautifulSoup

def clean_html(text):
    if pd.isna(text):
        return ""

    return BeautifulSoup(
        str(text),
        "html.parser"
    ).get_text(" ", strip=True)
```

```python
text = clean_html(jobs.loc[27, "job_description"])

print(text[:2000])
```

```python
doc = nlp(text)

matches = matcher(doc)

found_skills = sorted(
    set(
        doc[start:end].text.strip()
        for _, start, end in matches
    ),
    key=str.lower
)

print("Job title:", jobs.loc[27, "title"])
print("\nExtracted skills:")
print(found_skills)
```

```python
def extract_skills_from_doc(doc):
    matches = matcher(doc)

    found = set()

    for _, start, end in matches:
        skill = doc[start:end].text.strip()

        if skill:
            found.add(skill)

    return sorted(found, key=str.lower)
```

```python
descriptions = jobs["job_description"].fillna("").map(clean_html).tolist()

all_extracted_skills = []

for i, doc in enumerate(nlp.pipe(descriptions, batch_size=1000)):

    all_extracted_skills.append(
        extract_skills_from_doc(doc)
    )

    if (i + 1) % 10000 == 0:
        print(f"Processed {i + 1:,} / {len(descriptions):,}")
```

```python
jobs["extracted_skills"] = all_extracted_skills

print("Extraction complete.")
```

```python
def canonicalize_skill(skill):
    s = str(skill).strip().lower()

    aliases = {
        "agile": "Agile",
        "research": "Research",
        "ci / cd": "CI/CD",
        "ci/cd": "CI/CD",
        "excel": "Excel",
        "microservices": "Microservices",
        "software development": "Software Development",
        "java script": "JavaScript",
        "power bi": "Power BI",
        "jira": "Jira",
        "gcp": "GCP",
        "aws": "AWS",
        "azure": "Azure",
        "postgres": "PostgreSQL",
        "sql server": "SQL Server",
        "kafka": "Kafka",
        "apache kafka": "Kafka"
    }

    return aliases.get(s, str(skill).strip())
```

```python
jobs["extracted_skills"] = jobs["extracted_skills"].apply(
    lambda skills: sorted(
        set(canonicalize_skill(skill) for skill in skills),
        key=str.lower
    )
)
```

```python
extracted_counter = Counter()

for skill_list in jobs["extracted_skills"]:
    extracted_counter.update(skill_list)

extracted_frequency = pd.DataFrame(
    extracted_counter.items(),
    columns=["skill", "job_count"]
).sort_values(
    "job_count",
    ascending=False
)

print("Unique canonical skills:", len(extracted_frequency))

extracted_frequency.head(50)
```

```python
display_names = {
    "go": "Go",
    "optimization": "Optimization",
    "networking": "Networking",
    "data engineering": "Data Engineering",
    "software engineering": "Software Engineering"
}

jobs["extracted_skills"] = jobs["extracted_skills"].apply(
    lambda skill_list: sorted(
        {
            display_names.get(
                skill.lower(),
                skill
            )
            for skill in skill_list
        },
        key=str.lower
    )
)
```

```python
skills_df = pd.DataFrame({
    "skill_name": extracted_frequency["skill"].tolist()
})

skills_df.insert(
    0,
    "skill_id",
    range(1, len(skills_df) + 1)
)

print("Skills:", len(skills_df))
skills_df.head(20)
```

```python
skills_df.to_sql(
    "skills",
    engine,
    if_exists="replace",
    index=False
)

print("✅ skills table created")
```

```python
job_skill_rows = []

for _, row in jobs[
    ["job_id", "extracted_skills"]
].iterrows():

    for skill in row["extracted_skills"]:

        job_skill_rows.append({
            "job_id": row["job_id"],
            "skill_name": skill
        })

job_skills_df = pd.DataFrame(job_skill_rows)

print("Raw relationships:", len(job_skills_df))
```

```python
job_skills_df = job_skills_df.merge(
    skills_df,
    on="skill_name",
    how="inner"
)
```

```python
job_skills_df = job_skills_df[
    ["job_id", "skill_id"]
].drop_duplicates()

print(
    "Unique job-skill relationships:",
    len(job_skills_df)
)
```

```python
job_skills_df.to_sql(
    "job_skills",
    engine,
    if_exists="replace",
    index=False,
    chunksize=10000,
    method="multi"
)

print("✅ job_skills table created")
```
